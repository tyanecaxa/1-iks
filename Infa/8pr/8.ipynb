```python
import numpy as np
import matplotlib.pyplot as plt

# --- Данные ---------------------------------------------------------------
days_labels = [str(i+1) for i in range(14)]

# Температура днём и ночью (примерные значения, °C)
days = np.array([5.1, 6.3, 4.8, 7.2, 9.5, 8.1, 6.7, 
               5.9, 7.4, 8.8, 10.2, 9.1, 7.5, 6.2])
night = np.array([-1.2, 0.3, -2.1, 1.5, 3.8, 2.4, 0.9,
                 -0.4, 1.8, 3.2, 4.6, 2.9, 1.1, -0.3])

# Среднее значение днем и ночью
sr_d = np.mean(days)
sr_n = np.mean(night)

# --- Построение -----------------------------------------------------------

# размер в дюймах (ширина × высота). При dpi=100 это ~900×450 пикселей на экране.
plt.figure(figsize=(9, 4.5))

# добавление линий
plt.axhline(y=0, color='lightgray', linewidth=2, zorder=1)
plt.axhline(y=sr_n, color='lightblue', linewidth=2, zorder=1)
plt.axhline(y=sr_d, color='yellow', linewidth=2, zorder=1)

plt.plot(days_labels, night,
         color='blue',
         linewidth=2, 
         zorder=2,
         marker='s',
         markersize=6,
         label='Ночь')

plt.plot(days_labels, days,
        color='orange',
        linewidth=2,
        zorder=2,
        marker='o',
        markersize=6,
        label='День')

# --- Оформление -----------------------------------------------------------
plt.title('Температура воздуха в течение 2х недель', fontsize=13)
plt.xlabel('День')
plt.ylabel('Температура, °C')

plt.grid(True, linestyle='--', alpha=0.6)


plt.legend(loc='upper left')
plt.tight_layout()                          # Подгоняет отступы так, чтобы подписи осей и заголовок не обрезались по краям
plt.savefig('weather.png', dpi=150)     # Сохраняет текущую фигуру в файл
plt.show()

# Сколько дней ночью была температура ниже нуля
m0 = [i for i in night if i < 0]
print('Ночей с температурой ниже нуля:',len(m0))
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --- Данные ---------------------------------------------------------------
days_labels = [i+1 for i in range(14)]

# Температура днём и ночью (примерные значения, °C)
hour = np.array([4,5, 5.2, 5.8, 4.9, 6.1, 6.8, 7.2, 
               7.5, 8.1, 7.8, 8.5, 9.0, 9.3, 9.7])
energy = np.array([58, 52, 48, 55, 42, 38, 33,
                 29, 22, 25, 18, 14, 11, 8])

# Среднее значение днем и ночью
sr_d = np.mean(hour)
sr_n = np.mean(energy)

# --- Построение ------------------------------------------------------------
# Создание фигуры с двумя панелями
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 5))

# Левая панель
ax1.plot(range(len(hour)), hour, color='blue', linewidth=2)
ax1.axhline(y=sr_d, color='darkorange', linewidth=2, zorder=1)

ax1.set_title('Часы работы экрана')
ax1.set_xlabel('День')
ax1.set_ylabel('Экран, ч')
ax1.set_ylim(0, 11)
ax1.set_xticks(days_labels)
ax1.grid(True, linestyle='--', alpha=0.6)
ax1.legend()

# Правая панель
ax2.plot(days_labels, energy, color='red', linewidth=2)
ax2.axhline(y=sr_n, color='green', linewidth=2, zorder=1)

ax2.set_title('Процент заряда к концу дня')
ax2.set_xlabel('День')
ax2.set_ylabel('Заряд, %')
ax2.set_ylim(0, 100)
ax2.set_xticks(days_labels)
ax2.grid(True, linestyle='--', alpha=0.6)
ax2.legend()

fig.tight_layout()
fig.savefig('phone.png', dpi=150)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --- Данные ---------------------------------------------------------------

# Названия пиццерий и кол-во проданной пиццы
names = np.array(['Маргарита Мания', "Пепперони-Парад", "Гавайский Хаос",
                  "Четыре Сыра и Слёзы", "Веганский Бунт"])
sales = np.array([23, 45, 56, 78, 32])
x = np.arange(len(names))

winner = np.argmax(sales)
print("Пицца победитель: ", names[winner])

# --- Построение ------------------------------------------------------------
# Создание фигуры и диаграммы
fig, ax = plt.subplots(figsize=(10, 5))

ax.bar(x, sales, width=0.6, color='lightskyblue', edgecolor='purple')

ax.set_title('Битва пиццерий: продажи за неделю')
ax.set_xlabel('Пиццерия')
ax.set_ylabel('Продано пицц, шт.')

fig.tight_layout()
fig.savefig('pizzeria.png', dpi=150)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --- Данные ---------------------------------------------------------------

labels = np.array(['Спасает мир', "Восстанавливает суперсилы во сне", 
                   "Отдыхает, играя в кибер-квесты", "Другое"])
hours = np.array([8, 7, 5, 4])
x = np.arange(len(labels))

# --- Построение ------------------------------------------------------------
# Создание фигуры и диаграммы
fig, ax = plt.subplots(figsize=(7, 7))
ax.pie(hours, labels=labels, autopct='%1.1f%%', 
       colors=['papayawhip', 'turquoise', 'mediumpurple', 'gray'])

ax.set_title('День супергероя Алексея')

fig.tight_layout()
fig.savefig('superhero.png', dpi=150)
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# --- Данные ---------------------------------------------------------------
dist = np.array([100, 500, 1000, 3000, 5000, 10000])
ping = np.array([1.8, 5.9, 11.5, 32.5, 53.0, 104.2])

n = 1.5
c = 300000
v = c / n
P = np.array([((2 * i / v) * 1000) for i in dist])

# --- Построение ------------------------------------------------------------
plt.figure(figsize=(9, 4.5))

plt.plot(dist, ping, 
         linestyle='None',
         color='tomato',
         zorder=2,
         marker='o',
         markersize=4,
         label='Экспериментальный пинг')
plt.plot(dist, P,
         linestyle='-',
         color='mediumspringgreen', 
         linewidth=2,
         label='Теоретический пинг')

plt.title('Пинг и дистанция', fontsize=13)
plt.xlabel('Дистанция, км')
plt.ylabel('Пинг, мс')
plt.xticks(dist)
plt.yticks(ping)
plt.grid(True, linestyle='--', alpha=1)
plt.legend()

plt.tight_layout()
plt.savefig('ping_test.png', dpi=150)
plt.show()
```
