Задание 1 "Сердце"

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.animation import FuncAnimation

t = np.linspace(0, 2*np.pi, 250)
x = np.array([16*np.sin(t1)**3 for t1 in t])
y = np.array([13*np.cos(t2) - 5*np.cos(2*t2) 
              - 2*np.cos(3*t2) - np.cos(4*t2) for t2 in t])

fig, ax = plt.subplots(figsize=(9, 3))
ax.set_xlim(0, 4 * np.pi)
ax.set_ylim(-1.2, 1.2)

line, = ax.plot([], [], 'b-', lw=2)
dot,  = ax.plot([], [], 'ro', ms=8)

def update(frame):
    n = frame + 1
    line.set_data(x[:n], y[:n])         # ← срез
    dot.set_data([x[n-1]], [y[n-1]])
    return line, dot

ani = FuncAnimation(fig, update, frames=len(x), interval=30)
plt.show()
```
